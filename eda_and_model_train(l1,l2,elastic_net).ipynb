```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv("Algerian_forest_fires_dataset_clean.csv")
```

```python
df['Classes'] = np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
df.drop(['day','month','year'],axis=1,inplace=True)
```

```python
X = df.drop('FWI',axis=1)
y = df['FWI']
```

```python
X.corr()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(X.corr()
            ,annot=True)
```

```python
def multicoll(datset,threshold):
    cor = datset.corr()
    col = set()
    for i in range(len(cor.columns)):
        for j in range(i):
            if abs(cor.iloc[i][j])>threshold:
                c = cor.columns[i]
                col.add(c)
    return col
            
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
multicoll(X_train,0.9)
```

```python
multicoll(X_train,0.85)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
X_train_stand = sc.fit_transform(X_train)
X_test_stand = sc.transform(X_test)
```

```python
plt.figure(figsize=(12,10))
plt.subplot(1,2,1)
plt.xticks(rotation = 45)
sns.boxplot(data=X_train)
plt.subplot(1,2,2)
sns.boxplot(data = X_train_stand)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
from sklearn.linear_model import RidgeCV
from sklearn.linear_model import Lasso
from sklearn.linear_model import LassoCV
from sklearn.linear_model import ElasticNet
from sklearn.linear_model import ElasticNetCV
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
```

Linear model

```python
li  = LinearRegression()
li.fit(X_train_stand,Y_train)
y_pred = li.predict(X_test_stand)
mae = mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print(f"mae: {mae}")
print(f"score: {r2}")
```

##RIDGE l2

```python
r = Ridge()
r.fit(X_train_stand,Y_train)
y_pred = r.predict(X_test_stand)
mae = mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print(f"mae: {mae}")
print(f"score: {r2}")
```

LASSO L1

```python
l = Lasso()
l.fit(X_train_stand,Y_train)
y_pred = l.predict(X_test_stand)
mae = mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print(f"mae: {mae}")
print(f"score: {r2}")
```

ELASTIC NET

```python
e = ElasticNet()
e.fit(X_train_stand,Y_train)
y_pred = e.predict(X_test_stand)
mae = mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print(f"mae: {mae}")
print(f"score: {r2}")
```

RIDGE CV

```python
rc = RidgeCV()
rc.fit(X_train_stand,Y_train)
y_pred = rc.predict(X_test_stand)
mae = mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print(f"mae: {mae}")
print(f"score: {r2}")
print(f" alpha: {rc.alpha_}")
# print(f"all alphas: {rc.alphas_}")
rc.alphas
```

LASSO CV

```python
lc = LassoCV()
lc.fit(X_train_stand,Y_train)
y_pred = lc.predict(X_test_stand)
mae = mean_absolute_error(Y_test,y_pred)
r2 = r2_score(Y_test,y_pred)
print(f"mae: {mae}")
print(f"score: {r2}")
print(f" alpha: {lc.alpha_}")
print(f"alphas : {lc.alphas_}")
```
